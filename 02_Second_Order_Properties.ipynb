# Second-Order Properties of Time Series Model

**Definition**: Let $\{X_t\}$ be a time series model for the observed time series data $\{x_t\}$. **Second-order properties** of time series model $\{X_t\}$ are:

* **Mean Function** $\mu_t$ that depends on time, defined as:
$$
\mu_t = E[X_t], \, \forall t
$$

* **Variance Function** $\sigma^2_t$ that depends on time, defined as:
$$
\sigma^2_t = Var[X_t] = E[(X_t - \mu_t)^2], \, \forall t
$$

* **Autocovariance Function** $\gamma(t, s)$ that depends on time steps $t$ and $s$, defined as:
$$
\gamma(t, s) = Cov[X_t, X_s] = E[(X_t - \mu_t)(X_s - \mu_s)], \, \forall t, s
$$

* **Autocorrelation Function (ACF)** $\rho(t, s)$ that depends on time steps $t$ and $s$, defined as:
$$
\rho(t, s) = Corr[X_t, X_s] = \dfrac{Cov[X_t, X_s]}{\sqrt{Var[X_t] \cdot Var[X_s]}} = \dfrac{\gamma(t, s)}{\sqrt{\gamma(t) \cdot \gamma(s)}}, \, \, \forall t, s
$$

### Example 1. IID Gaussian random variables

**Example**: Let $\{X_t\}_{t \in \mathbf{N}}$ be a sequence of independent and identically distributed (i.i.d.) random variables such that:
$$
X_t \sim \mathcal{N}(0, 1^2)
$$

$\{X_t\}_{t \in \mathbf{N}}$ is stochastic process.

Let's determen **second-order properties** of time series model $\{X_t\}$

**Mean function**

$$
\mu_t = E[X_t] = 0, \, \, \forall t
$$
Time series model $\{X_t\}$ has **constant zero-mean function**

**Variance function**

$$
\sigma^2_t = Var[X_t] = 1, \, \, \forall t
$$
Time series model $\{X_t\}$ has **constant variance function**

**Autocovariance function**

$$
\forall t \ne s: \, \, \gamma(t, s) = Cov[X_t, X_s] = E[(X_t - \mu_t)(X_s - \mu_s)]=E[X_t \cdot X_s] \overset{\text{independence}}{=}=E[X_t] \cdot E[X_s] = 0, \\
\forall t: \, \, \gamma(t, t) = Cov[X_t, X_t] = E[(X_t - \mu_t)(X_t - \mu_t)]=Var[X_t] = \sigma_t^2 = 1
$$

Autocovariance of time series model $\{X_t\}$ is equal to $\gamma(t,s)=0$ for $t \ne s$ which means that there is no linear dependance between elements $X_t$ and $X_s$.

**Autocorrelation function**

$$
\forall t \ne s: \, \, \rho(t, s) = Corr[X_t, X_s] = \dfrac{\gamma(t, s)}{\sqrt{\sigma^2_t \cdot \sigma^2_s}} = 0 \\
\forall t: \, \, \rho(t, t) = \dfrac{\gamma(t, t)}{\sqrt{\sigma^2_t \cdot \sigma^2_t}} = \dfrac{\sigma^2_t}{\sigma^2_t} = 1
$$

Autocorrelation of time series model $\{X_t\}$ is equal to $\rho(t,s)=0$ for $t \ne s$. $\rho(t,t)=1$ holds for each time series model.

Let's sample $n=100$ data points from process $\{X_t\}$ and plot it.

```python
import numpy as np
from matplotlib import pyplot as plt

# Generate time series data by sampling from the i.i.d. Gaussian distribution
time_series = np.random.normal(0, 1, 100)

time = np.arange(100) + 1

# Plot the time series
plt.figure()
plt.plot(time, time_series, label='Time series data sampled from the i.i.d. Gaussian distribution')
plt.axhline(y=0, color='red', linestyle='--', linewidth=1, label=f'Mean function')

plt.xlabel('Time')
plt.ylabel('Value')
plt.legend()
plt.grid(True)
plt.show()
```

Red line on the plot is mean function. We can see that sampled time series is centered around the mean function.

Variation function is constant, so spred of the datapoints around the red line is constant.

### Example 2. Simple Random Walk

**Example**: Let $\{Z_t\}_{t \in \mathbf{N}}$ be a sequence of independent and identically distributed (i.i.d.) random variables such that:
$$
P\{Z_t=1\} = p \\
P\{Z_t=-1\} = 1 - p
$$

The **simple random walk** $\{S_t\}_{t \in \mathbf{N_0}}$ is stochastic process defines by the partial sums:

$$
S_0 = 0 \\
S_t = Z_1 + Z_2 + ... + Z_t = \sum_{i=1}^t Z_i
$$

Sequence $\{Z_t\}_{t \in \mathbf{N}}$ is stochastic process itself with the following second-order properties:

$$
\mu_Z = E[Z_t] = 1 \cdot p + (-1) \cdot (1-p) = p-1+p = 2p-1  \\
\sigma_{Z, t}^2 = Var[Z_t] = E[(Z_t - \mu_t)^2] = E[(Z_t - 2p + 1)^2] = (1 - 2p + 1)^2 \cdot p + (-1 - 2p + 1)^2 \cdot (1-p) = 4p(1-p) \\
$$

Because elements are independent for $t \ne s$ we have:
$$
\gamma_Z (t, s) = 0 \\
\rho_Z (t, s) = 0 \\
$$

For any $t$ we have:
$$
\gamma_Z (t, t) = \sigma_t^2 = 4p(1-p)\\
\rho_Z (t, t) = 1 \\
$$

Let's determen **second-order properties** of time series model $\{S_t\}$

**Mean function**

$$
\mu_0 = E[S_0] = 0, \\
\mu_t = E[S_t] = E[\sum_{i=1}^t Z_i] = \sum_{i=1}^t E[Z_i] = t \cdot \mu_Z = t \cdot (2p-1), \, \, \forall t \ge 1
$$
Mean function of simple random walk $\{S_t\}$ is a linear function of time. Mean function increases if $p>\frac{1}{2}$ and decreases if $p<\frac{1}{2}$.

**Variance function**

$$
\sigma^2_t = Var[S_t] = Var\left[\sum_{i=1}^t Z_i \right] \overset{\text{independence}}{=} \sum_{i=1}^t Var\left[Z_i \right] = t \cdot 4p(1-p) , \, \, \forall t
$$
Variance function of simple random walk is linear function of time. Because $4p(1-p)>0, \, \forall p$, variance function is increasing function.

**Autocovariance function**

$$
\forall t < s: \, \, \gamma(t, s) = Cov[S_t, S_s] = Cov[\sum_{i=1}^t Z_i, \sum_{j=1}^s Z_j] =
\sum_{i=1}^t \sum_{j=1}^s Cov[Z_i, Z_j] =
\sum_{i, j=1}^t Cov[Z_i, Z_j] + \sum_{i=1}^t \sum_{j=t+1}^{s} Cov[Z_i, Z_j] =
\sum_{i=j=1}^t Cov[Z_i, Z_i] + \sum_{i,j=1 \\ i \ne j}^t Cov[Z_i, Z_i] +
\sum_{i=1}^t \sum_{j=t+1}^{s} Cov[Zi, Zj] =
\sum_{i=j=1}^t \sigma_{Z,i}^2 = t \cdot 4p(1-p)
$$

**Autocorrelation function**

$$
\forall t < s: \, \, \rho(t, s) = Corr[X_t, X_s] = \dfrac{\gamma(t, s)}{\sqrt{\sigma^2_t \cdot \sigma^2_s}} = \dfrac{t \cdot 4p(1-p)}{\sqrt{(t \cdot 4p(1-p)) \cdot (s \cdot 4p(1-p))}} = \dfrac{t \cdot 4p(1-p)}{4p(1-p)\sqrt{t \cdot s}} = \sqrt{\dfrac{t}{s}} \\
\forall t: \, \, \rho(t, t) = 1
$$
If we fix $t$ and let $s \ge t$ increase autocorrelation function $\sqrt{\dfrac{t}{s}}$ decreases. That means that for fixed $t$ linear dependance of random variables $S_t$ and $S_s$ is smaller as $s \rightarrow \infty$.

Let's sample $n=1000$ data points from **symmetric simple walk** $\{S_t\}$ where $p=\frac{1}{2}$ and plot it.

```python
# Generate Z_t
z = np.random.choice((-1, 1), size = 1000)
s = np.cumsum(z)
t = np.arange(1000) + 1

# Plot the time series
plt.figure()
plt.plot(t, s, label='Symmetric simple random walk')
plt.axhline(y=0, color='red', linestyle='--', linewidth=1, label=f'Mean function')

plt.xlabel('Time')
plt.ylabel('Value')
plt.legend()
plt.grid(True)
plt.show()
```

Red line on the plot is mean function. We can see that sampled time series is more less centered around the mean function.

Variation function is increasing linear function of time, so spred of the datapoints around the red line gets larger as $t$  gets larger.

For symetric random walk $Var[X_t]=t$. This is why the walk tends to wander farther from zero as time passes.
